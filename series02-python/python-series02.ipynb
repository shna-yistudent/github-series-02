## Mathematical model
A sinusodial signal can be described by

$$
s(t)=A\sin(2\pi f t + \phi)
$$

where:

-$A$ is the amplitude
-$f$ is the frequency in Hz
-$t$ is time in seconds
-$\phi$ is the phase

# python series 012- sinusodial signals
introduces the generation and visualisationof a sinusodial signal suing python

```python
import numpy as np
import matplotlib.pyplot as plt

# create the time array
t = np.linspace(0, 1, 1000)  # time from 0 to 1 second, 1000 points

# create the sinusoidal signal
f = 1  # frequency in Hz
s = np.sin(2 * np.pi * f * t)  # sine wave

#create the cosine signal
c = np.cos(2 * np.pi * f * t)  # cosine wave

#plot both signals
plt.plot(t, s, label='Sine')
plt.plot(t, c, label='Cosine')

# Add the title and labels
plt.title('sinusoidal signal')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')

# Add a legend
plt.legend()

# add grid
plt.grid(True)

#save the plot as a PDF file
plt.savefig('sinusoidal_signal.pdf', format='pdf')


# display the figure
plt.show()
```
