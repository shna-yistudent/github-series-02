## Mathematical model
A sinusodial signal can be described by

$$
s(t)=A\sin(2pi f t + \phi)
$$

where:

-$A$ is the amplitude
-$f$ is the frequency in Hz
-$t$ is time in seconds
-$\phi$ is the phase

```python
import numpy as np
import matplotlib.pyplot as plt

# create the time array
t = np.linspace(0, 1, 1000)  # time from 0 to 1 second, 1000 points

# create the sinusoidal signal
f = 5  # frequency in Hz
s = np.sin(2 * np.pi * f * t)  # sine wave

#plot the signal
plt.plot(t, s)

# Add the title and labels
plt.title('sinusoidal signal')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')

# add grid
plt.grid(True)

# save the figure
plt.savefig('sinusoidal_signal.png')

# display the figure
plt.show()
```
